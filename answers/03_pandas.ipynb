## C1 — Series and DataFrame by hand

## C10 — Derived columns

```python
invalid_marks = ~df["marks"].between(0,100)
print("C10 - Impossible marks:")
print(df.loc[invalid_marks, ["student_id", "name", "marks"]])
```

```python
df.loc[invalid_marks, "marks"] = np.nan
df = df.dropna(subset=["marks"])
df["passed"]=df["marks"]>=40
df["grade"]=np.where(
    df["marks"]>=85,"A",
    np.where(
        df["marks"]>=70,"B",
        np.where(
            df["marks"]>=40,"C","F")
    )
)
pass_rate = df["passed"].astype(int).mean() * 100
print("C10 - Pass rate:", pass_rate, "%")
print("C10 - Grade counts:")
print(df["grade"].value_counts())
```

```python
# Impossible marks were treated as missing and removed as they cannot represent valid exam marks.
```

## C11 — Types and memory

```python
print("C11 - Memory before:")
print(df[["subject", "city"]].memory_usage(deep=True))
```

```python
mem_before = df[["subject", "city"]].memory_usage(deep=True).sum()
```

```python
mem_before
```

```python
df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")
print("C11 - Memory after:")
print(df[["subject", "city"]].memory_usage(deep=True))
```

```python
mem_after = df[["subject", "city"]].memory_usage(deep=True).sum()
```

```python
mem_after
```

```python
mem_saved = ((mem_before - mem_after) / mem_before)*100
print("C11 - Memory saved:", mem_saved,"%")
```

```python
# Category is cheaper because repeated text values are stored as category codes.
# It would not help when most values are unique.
```

## C12 — Aggregation and a summary table
