```python
import numpy as np
np.random.seed(7)
cities = np.array(["Kathmandu", "Pokhara", "Biratnagar", "Lalitpur", "Birgunj"])
days = np.array([f"D{i:02d}" for i in range(1, 15)])
temps = np.random.randint(14, 36, size=(5, 14)).astype(float) # °C, 5 cities × 14 days
humidity = np.random.randint(40, 95, size=(5, 14)).astype(float) # %
temps[2, 5] = np.nan # sensor outage
temps[4, 11] = np.nan
```

## B1 — Array anatomy and reshaping

```python
print("B1a - temps shape:", temps.shape)
print("B1b - temps dimension:", temps.ndim)
print("B1c - temps data type:", temps.dtype)
print("B1d - temps size:", temps.size)
```

```python
nums = np.arange(70)
```

```python
reshaped_explicit = nums.reshape(5, 14)
reshaped_a = nums.reshape(5, -1)

print("B1e - Reshaped with explicit dimensions:", reshaped_explicit)
print("\nB1f - Reshaped using -1:", reshaped_a)

print("\nB1g - Two results identical:", np.array_equal(reshaped_explicit, reshaped_a))
```

```python
flattened = reshaped_explicit.flatten()
print("B1h - Flattened array:", flattened)
```

```python
# -1 lets NumPy finds out the missing dimension automatically.
# Reshaping does not copy the data as it changes the way we view the data
```

## B3 — Boolean masks

```python
mask = temps > 30

print("B3a - Readings above 30 per city:", np.sum(mask, axis=1))

kathmandu = temps[0] > 30
print("B3b - Kathmandu days above 30:", days[kathmandu])
```

## B5 — Vectorization vs. Python loop

```python
temps[0, 0] = 20
print(temps[0, :3])
```

```python
import time

temps_loop = np.empty_like(temps)
rows, cols = temps.shape
for r in range(rows):
    for c in range(cols):
        temps_loop[r, c] = (temps[r,c] * 9 / 5) + 32

temps_vector = (temps * 9 / 5) + 32
print("B5a - Vectorized and loop results match:", np.allclose(temps_loop, temps_vector, equal_nan=True))

large_arr = np.random.randint(14, 26, size = 1_000_000).astype(float)

t0 = time.perf_counter()
large_loop = [x * 9 / 5 + 32 for x in large_arr]
t_loop = time.perf_counter() - t0

t0 = time.perf_counter()
large_vector = (large_arr * 9 / 5) + 32
t_vector = time.perf_counter() - t0

speedup = t_loop / t_vector

print(f"B5b - Loop time: {t_loop:.5f}s")
print(f"B5b - Vectorized time: {t_vector:.5f}s")
print(f"B5b - Vectorized code is roughly {speedup:.2f}x faster")

# The loop takes extra time for repeated Python-level iteration, indexing, and assignment.
```

## B6 — Broadcasting

```python
city_avg = np.nanmean(temps, axis=1, keepdims=True)

anomalies = temps - city_avg

print("B6a - Anomaly shape:", anomalies.shape)
print("B6a - Confirm shape (5, 14)?", anomalies.shape == (5, 14))

sensor_correction = np.array([
    0.2, -0.1, 0.3, 0.0, -0.2, 0.1, 0.2,
    -0.1, 0.0, 0.3, -0.2, 0.1, 0.0, -0.1
])

corrected_temps = temps + sensor_correction

print("\nB6b - Corrected temps shape:", corrected_temps.shape)
print("B6b - Corrected temperatures:", np.round(corrected_temps[:2, :4], 2))

try:
    wrong_result = temps + np.array([1, 2, 3, 4, 5])
except ValueError as e:
    print("\nB6c - ValueError:", e)

# Broadcasting aligns shapes from the right, so (14,) matches (5, 14) in (b), but (5,) does not in (c) causing dimension mismatch.
```

## B7 — Matrix multiplication

```python
quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])

print("B7a - quantities * prices:", quantities * prices)
print("B7a - quantities @ prices:", quantities @ prices)

mean_temp_per_day = np.nanmean(temps, axis=0)
mean_humidity_per_day = np.nanmean(humidity, axis=0)

features = np.column_stack((mean_temp_per_day, mean_humidity_per_day))

weights = np.array([0.6, -0.1])

comfort_score = features @ weights

print("\nB7b - Features shape:", features.shape)
print("B7b - Weights shape:", weights.shape)
print("B7b - Comfort score shape:", comfort_score.shape)
print("B7b - Daily comfort scores:", np.round(comfort_score, 2))

# * multiplies corresponding elements, while @ calculates the dot product.
# The inner dimensions must match so each feature can be multiplied by its corresponding weight.
```

## B8 — Aggregating along an axis

```python
city_mean = np.nanmean(temps, axis=1)
city_max = np.nanmax(temps, axis=1)
city_min = np.nanmin(temps, axis=1)
city_std = np.nanstd(temps, axis=1)

print("B8a - Mean per city:", city_mean)
print("Shape:", city_mean.shape)

print("B8a - Max per city:", city_max)
print("Shape:", city_max.shape)

print("B8a - Min per city:", city_min)
print("Shape:", city_min.shape)

print("B8a - Std per city:", city_std)
print("Shape:", city_std.shape)

day_mean = np.nanmean(temps, axis=0)
day_max = np.nanmax(temps, axis=0)

print("\nB8b - Mean per day:", day_mean)
print("Shape:", day_mean.shape)

print("B8b - Max per day:", day_max)
print("Shape:", day_max.shape)

hottest_day_index = np.nanargmax(temps, axis=1)
hottest_days = days[hottest_day_index]

print("\nB8c - Hottest day for each city:", hottest_days)
print("Shape:", hottest_days.shape)

# axis=0 collapses the cities, while axis=1 collapses the days.
```

## B9 — Reproducible randomness and a train/test split

```python
flat_temps = temps.flatten()
np.random.seed(42)
indices = np.arange(len(flat_temps))
np.random.shuffle(indices)

split = int(0.8 * len(indices))
train_index = indices[:split]
test_index = indices[split:]

print("B9a - Train shape:", train_index.shape)
print("B9a - Test Shape:", test_index.shape)
print("B9a - Total samples sum to 70:", len(train_index) + len(test_index) == 70)
print("B9a - Share no index:", len(np.intersect1d(train_index, test_index)) == 0)

np.random.seed(42)
again = np.arange(len(flat_temps))
np.random.shuffle(again)

train2 = again[:split]
test2 = again[split:]

assert np.array_equal(train_index, train2)
assert np.array_equal(test_index, test2)
print("\nB9b - Assert passed. Hence, the split is identical.")

# Shuffling mixes the samples before splitting, while the seed makes the same split repeatable.
```

## B10 — np.where , argmax , argmin

```python
label = np.where(temps > 30, "hot", np.where(temps >= 22, "mild", "cold"))
print("B10a - Label's grid shape:", label.shape)
print("B10a - Sample:", label[:2, :5]) 

scores = np.array([0.08, 0.62, 0.21, 0.09])
labels = np.array(["rain", "clear", "cloudy", "fog"])

predicted_class = labels[np.argmax(scores)]
print(f"\nB10b- Predicted label: {predicted_class}")

# A classifier needs argmax to get the position of the highest score and select its corresponding label.
# max() returns only the float score instead of the label index.
```
